```python
def fix_encoding(text):
    if pd.isna(text):
        return text
    text=str(text)
    try:
        return text.encode("latin1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return text
```

```python
def fix_multiple_encoding(text, max_attempts=5):
    if pd.isna(text):
        return text

    text = str(text)

    for _ in range(max_attempts):
        try:
            fixed = text.encode("latin1").decode("utf-8")

            # Stop if another conversion doesn't change anything
            if fixed == text:
                break

            text = fixed

        except (UnicodeEncodeError, UnicodeDecodeError):
            break

    return text
```

```python
df["combined_text"] = (
    df["name"].fillna("") + " " +
    df["cuisines"].fillna("") + " " +
    df["rest_type"].fillna("") + " " +
    df["location"].fillna("") + " " +
    df["restaurant_profile"].fillna("")
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

tfidf_matrix = tfidf.fit_transform(df["combined_text"])

print("TF-IDF shape:", tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(tfidf_matrix)

print("Similarity matrix shape:", cosine_sim.shape)
```

```python
indices = pd.Series(
    df.index,
    index=df["name"].str.lower()
).drop_duplicates()

print(indices.head())
```

```python
restaurant = "Jalsa"
matches = df.index[
    df["name"].str.lower() == restaurant.lower()
]
print("Matching rows:", len(matches))
idx = int(matches[0])

# Get similarity scores for that restaurant
similarity_scores = cosine_sim[idx].ravel()

# Create (index, score) pairs
similarity_scores = list(enumerate(similarity_scores))

# Sort by similarity
similarity_scores = sorted(
    similarity_scores,
    key=lambda x: float(x[1]),
    reverse=True
)

# Display top 10
for i, score in similarity_scores[:10]:
    print(df.loc[i, "name"], round(float(score), 4))
```

```python
restaurant = "Jalsa"

matches = df.index[
    df["name"].str.lower() == restaurant.lower()
]

idx = int(matches[0])

similarity_scores = cosine_sim[idx].ravel()

similarity_scores = [
    (i, float(score))
    for i, score in enumerate(similarity_scores)
    if df.loc[i, "name"].lower() != restaurant.lower()
]

similarity_scores = sorted(
    similarity_scores,
    key=lambda x: x[1],
    reverse=True
)

for i, score in similarity_scores[:10]:
    print(
        df.loc[i, "name"],
        "| similarity:",
        round(score, 4)
    )
```

```python
# Get similar restaurants
restaurant = "Jalsa"

matches = df.index[
    df["name"].str.lower() == restaurant.lower()
]

idx = int(matches[0])

similarity_scores = cosine_sim[idx].ravel()

recommendations = pd.DataFrame({
    "index": range(len(df)),
    "similarity_score": similarity_scores
})

# Add restaurant information
recommendations["name"] = df["name"].values
recommendations["cuisines"] = df["cuisines"].values
recommendations["location"] = df["location"].values
recommendations["rate"] = df["rate"].values
recommendations["votes"] = df["votes"].values

# Remove the input restaurant itself
recommendations = recommendations[
    recommendations["name"].str.lower() != restaurant.lower()
]

# Sort by similarity
recommendations = recommendations.sort_values(
    "similarity_score",
    ascending=False
)

# Keep only one row per restaurant
recommendations = recommendations.drop_duplicates(
    subset=["name"],
    keep="first"
)

recommendations.head(10)
```
